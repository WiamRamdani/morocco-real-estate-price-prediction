```python
import pandas as pd

# Charger le fichier CSV
df = pd.read_csv("Avito_local.csv")
```

```python
import mysql.connector

# Connexion à la base de données
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",  # ou ton mot de passe XAMPP
    database="avito_db"
)
cursor = conn.cursor()

# Nettoyage des noms de colonnes
columns = [col.replace(" ", "_").replace("-", "_").replace("é","e") for col in df.columns]

# Créer dynamiquement une table avec des types VARCHAR(255) par défaut
table_name = "Avito_local"

create_query = f"""
CREATE TABLE IF NOT EXISTS {table_name} (
    {', '.join(f'`{col}` VARCHAR(255)' for col in columns)}
);
"""
cursor.execute(create_query)
conn.commit()
```

```python
# Mettre à jour les noms dans le DataFrame pour correspondre à la table
df.columns = columns
df = df.fillna("")

insert_query = f"""
INSERT INTO {table_name} ({', '.join(f'`{col}`' for col in columns)})
VALUES ({', '.join(['%s'] * len(columns))})
"""

for _, row in df.iterrows():
    cursor.execute(insert_query, tuple(row))

conn.commit()
cursor.close()
conn.close()
```
